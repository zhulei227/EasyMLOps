```python
import pandas as pd
data=pd.read_csv("./data/demo.csv")
data.head()
target=data["Survived"]
del data["Survived"]
```

```python
from easymlops import TablePipeLine
from easymlops.table.preprocessing import *
from easymlops.table.ensemble import Parallel
from easymlops.table.strategy import HillClimbingStackingRegression,AutoResidualRegression
```

```python
pipeline=TablePipeLine()
pipeline-FixInput()-FillNa()-Parallel([
    AutoResidualRegression(y=target,pred_name="p1"),
    AutoResidualRegression(y=target,pred_name="p2"),
    AutoResidualRegression(y=target,pred_name="p3")
])-HillClimbingStackingRegression(y=target,eval_function="mae",eval_function_direct="minimize")
```

```python
pipeline.fit(data)
```

```python
pipeline.transform(data)
```

```python
pipeline.transform_single({})
```

```python
pipeline.transform(data).dtypes
```

```python
pipeline.auto_test(data)
```
