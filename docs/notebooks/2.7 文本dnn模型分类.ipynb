```python
import pandas as pd
data=pd.read_csv("./data/demo2.csv",encoding="gbk")
data.head(5)
```

```python
data.shape
```

```python
x=data[["review"]]
y=data["label"]

x_train=x[:7000]
x_test=x[7000:]
y_train=y[:7000]
y_test=y[7000:]

del data,x,y
```

### 训练

```python
from easymlops import NLPPipeline
from easymlops.table.preprocessing import *
from easymlops.nlp.preprocessing import *
from easymlops.nlp.text_classification import TextCNNClassification,TextRNNClassification,HANClassification
```

```python
nlp=NLPPipeline()
nlp.pipe(FixInput())\
   .pipe(FillNa(cols=["review"]))\
   .pipe(Lower())\
   .pipe(ExtractChineseWords())\
   .pipe(ExtractJieBaWords())\
   .pipe(VocabIndex())\
   .pipe(TextCNNClassification(y=y_train,col="review"))

x_test_new=nlp.fit(x_train).transform(x_test)
x_test_new.head(5)
```
